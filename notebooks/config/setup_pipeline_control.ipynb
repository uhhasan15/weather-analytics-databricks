```python
# Create a pipeline state table instead of relying on the bronze delta table (e.g MAX(request_end_date)) for most-recent date
# This reinforces clarity, and is more efficient in filling unusual > 1 day-length gaps in the bronze table

spark.sql("""
CREATE SCHEMA IF NOT EXISTS weather_project.control
""")

spark.sql("""
CREATE TABLE IF NOT EXISTS weather_project.control.pipeline_state
(
    pipeline_name STRING,
    source STRING,
    model STRING,
    last_successful_data_date DATE,
    last_successful_ingestion_id STRING,
    updated_at TIMESTAMP
)
USING DELTA
""")
```
